# BRIGHTMART SALES FORECAST PROJECT

## Task 1 - Create Variables

```python
product_name = "Wireless Mouse"
selling_price = 12500
units_sold = 180
weekly_sales_target  = 200
forecast_growth_rate = 15
current_stock = 250
minimum_stock_level = 120
product_available = True
```

## Task 2 — Calculate Total Weekly Revenue

```python
weekly_revenue = selling_price * units_sold
print(weekly_revenue)
```

## Task 3 — Forecast Next Week's Sales

```python
units_sold = 180
forecast_growth_rate = 15

forecasted_sales = units_sold + (units_sold * forecast_growth_rate / 100)
print("Forecasted Sales is" , forecasted_sales)
```

## Task 4 — Forecast Next Week's Revenue

```python
forecasted_revenue = forecasted_sales * selling_price
print(forecasted_revenue)
```

## Task 5 — Update Current Stock

```python
current_stock -= units_sold
print(current_stock)
```

## Task 6 — Determine if the Sales Target Was Achieved

```python
sales_target_achieved = units_sold >= weekly_sales_target
print(sales_target_achieved)
```

## Task 7 — Check if Restocking is Required

```python
restocking_required = current_stock < minimum_stock_level

print(restocking_required)
```

## Task 8 - Promotion Eligibility

```python
promotion_eligible = product_available and sales_target_achieved

print(promotion_eligible)
```

## Task 9 - Determine if Demand is High

```python
high_demand_forecast = (
    forecasted_sales > 200
    or forecasted_revenue > 2500000
)

print(high_demand_forecast)
```

## Task 10 - Weekly Sales Report

```python
report = f"""
=================================================
        BRIGHTMART SALES FORECAST REPORT
=================================================

Product Name           : {product_name}

Selling Price          : ₦{selling_price:,.2f}

Units Sold This Week   : {units_sold}

Weekly Revenue         : ₦{weekly_revenue:,.2f}

Forecast Growth Rate   : {forecast_growth_rate}%

Forecast Sales         : {forecasted_sales:.0f} Units

Forecast Revenue       : ₦{forecasted_revenue:,.2f}

Current Stock          : {current_stock} Units

Sales Target Achieved  : {sales_target_achieved}

Restocking Required    : {restocking_required}

Product Available      : {product_available}

Promotion Eligible     : {promotion_eligible}

High Demand Forecast   : {high_demand_forecast}

=================================================
"""

print(report)
```
